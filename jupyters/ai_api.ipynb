```python
# Add ID ... wowwwwww
import json
import os
import re

from pathlib import Path

directory = Path("/Users/anria/ab_python/fix search/justia/data_full/")
writedir = "/Users/anria/ab_python/fix search/justia/data4/"

for read_file in directory.glob("*.json"):
    write_file = writedir + read_file.name
    # print(read_file.name)
    with open(read_file) as file:
        json_data = json.load(file)
        json_data["id"] = re.sub(r'[^a-zA-Z0-9 ]', '', read_file.name.replace("json", "" ).replace(" ", "") )  # give yourself a chance to re-index reliably. What a thing to forget :D
        with open(write_file, "w") as wfile:
            json.dump(json_data, wfile, indent=4, sort_keys=True)


```

```python
import json
import os
from pathlib import Path
import shutil


directory = Path("/Users/anria/ab_python/django/adoption_cases/")
file_list = directory.glob("*.json")
writedir = "/Users/anria/ab_python/django/data4/"
i=0
# Loop through all text files
for read_file in file_list:
    if read_file.name.lower().find("_in re ") > 0:
        i+=1
        write_file = writedir + read_file.name
        print(i, "R=", read_file )
        shutil.copy(read_file, write_file)
```

```python
import ollama
from ollama import chat

model_name = 'FableForge-AI/nexus-security'

stream = chat(
    model=model_name,
    messages=[{'role': 'user', 'content': 'write a very very short poem by a person Considering putting out cigarette on herself.'}],
    stream=True,
)

for chunk in stream:
    print(chunk.message.content, end='', flush=True)

```
